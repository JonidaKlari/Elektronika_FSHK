## 1. Qëllimi i Punës Laboratorike
Qëllimi i këtij eksperimenti është studimi i funksionimit të **pjesëtuesit të tensionit**, krahasimi i sjelljes së qarqeve me rezistorë në rangun **kΩ** dhe **MΩ**, si dhe analizimi i ndikimit të rezistencës së ngarkesës (**R3 / RL**) mbi tensionin e daljes ($V_{out}$) dhe raportin e tensionit ($U_{out}/U_{in}$).

## 2. Bazat Teorike
* **Pjesëtuesi i Tensionit pa ngarkesë:** Për dy rezistorë $R_1$ dhe $R_2$ në seri të ushqyer me tension $V_{in}$, tensioni i daljes në skajet e rezistorit të poshtëm jepet nga:
  $$V_{out} = V_{in} \cdot \frac{R_2}{R_1 + R_2}$$
* **Raporti i Tensionit ($U_{out}/U_{in}$):** Tregon pjesën e tensionit hyrës që shfaqet në dalje; për dy rezistorë të afërt në vlerë ($R_1 \approx R_2$), raporti teorik pa ngarkesë është afërsisht $0.5$.
* **Efekti i Rezistorëve të Mëdhenj (MΩ) dhe Ngarkesës ($R_3 / R_L$):** Kur rezistorët kanë vlera në **MΩ**, rezistenca e brendshme e multimetrit apo e ngarkesës mund të ndikojë dukshëm në tensionin e matur. Ndërsa kur lidhet një ngarkesë $R_3 / R_L$ me vlera në rritje ($46\,\Omega$ deri në $1000\,\Omega$), tensioni i daljes rritet gradualisht me rritjen e rezistencës së ngarkesës.

## 4. Vizualizimi Grafik i Rezultateve
Të tre grafikët shfaqen bashkë në një figurë të vetme: (1) tensioni i daljes $V_{out}$ (V) për të gjitha matjet, (2) varësia e $V_{out}$ (V) nga rezistenca $R_3 / R_L$ ($\Omega$) në Eksperimentin 3, dhe (3) krahasimi i raportit $U_{out}/U_{in}$ për Eksperimentin 1 dhe 2.

```python
import matplotlib.pyplot as plt

eksperimenti = [
    "Eksperimenti 1",
    "Eksperimenti 2",
    "Eksperimenti 3 - Matja 1",
    "Eksperimenti 3 - Matja 2",
    "Eksperimenti 3 - Matja 3",
    "Eksperimenti 3 - Matja 4",
    "Eksperimenti 3 - Matja 5",
]

vout_val = [5.59, 4.98, 0.1600, 0.3480, 0.7656, 1.6786, 3.4783]
r3_rl_ohm = [46, 100, 220, 470, 1000]
vout_exp3 = [0.1600, 0.3480, 0.7656, 1.6786, 3.4783]

exp_12 = ["Eksperimenti 1", "Eksperimenti 2"]
uout_uin_12 = [0.4956, 0.4415]

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

bars1 = axes[0].bar(
    eksperimenti, vout_val, color="#58D68D", edgecolor="black", width=0.55
)
axes[0].set_title(
    "Tensioni i Daljes Vout (V) sipas Matjes", fontsize=11, fontweight="bold"
)
axes[0].set_xlabel("Eksperimenti")
axes[0].set_ylabel("Vout (V)")
axes[0].tick_params(axis="x", rotation=25)
axes[0].grid(axis="y", linestyle="--", alpha=0.5)
for bar, val in zip(bars1, vout_val):
  axes[0].annotate(
      f"{val:.4f} V",
      xy=(bar.get_x() + bar.get_width() / 2, bar.get_height()),
      xytext=(0, 4),
      textcoords="offset points",
      ha="center",
      va="bottom",
      fontsize=8,
  )

axes[1].plot(
    r3_rl_ohm,
    vout_exp3,
    marker="o",
    color="#2E86C1",
    linewidth=2,
    markersize=6,
)
axes[1].set_title(
    "Eksperimenti 3: Vout (V) sipas R3 / RL (Ω)", fontsize=11, fontweight="bold"
)
axes[1].set_xlabel("R3 / RL (Ω)")
axes[1].set_ylabel("Vout (V)")
axes[1].grid(True, linestyle="--", alpha=0.5)
for xv, yv in zip(r3_rl_ohm, vout_exp3):
  axes[1].annotate(
      f"{yv:.4f} V",
      xy=(xv, yv),
      xytext=(0, 7),
      textcoords="offset points",
      ha="center",
      fontsize=8,
  )

bars2 = axes[2].bar(
    exp_12,
    uout_uin_12,
    color=["#5DADE2", "#EC7063"],
    edgecolor="black",
    width=0.4,
)
axes[2].set_title("Raporti Uout / Uin", fontsize=11, fontweight="bold")
axes[2].set_ylabel("Uout / Uin")
axes[2].set_ylim(0, 0.62)
axes[2].grid(axis="y", linestyle="--", alpha=0.5)
for bar, val in zip(bars2, uout_uin_12):
  axes[2].annotate(
      f"{val:.4f}",
      xy=(bar.get_x() + bar.get_width() / 2, bar.get_height()),
      xytext=(0, 4),
      textcoords="offset points",
      ha="center",
      va="bottom",
      fontsize=8.5,
  )

plt.tight_layout()
plt.show()
```

## 5. Analiza e Gabimeve
* **Toleranca e rezistorëve dhe vlerat nominale:** Edhe kur rezistorët zgjidhen me vlera të përafërta (p.sh., $2.22\text{ k}\Omega$ dhe $2.18\text{ k}\Omega$ në Eksperimentin 1, ose $3.212\text{ M}\Omega$ dhe $3.250\text{ M}\Omega$ në Eksperimentin 2), devijimet nga vlera nominale dhe tolerancat ndikojnë në tensionin e daljes.
* **Ndikimi i rezistencës së brendshme të pajisjeve dhe kontaktet në breadboard:** Në vlera të larta rezistence ($\text{M}\Omega$), çdo rrjedhje apo rezistencë e brendshme e multimetrit ose kontakteve në *breadboard* ul tensionin e matur në dalje ($4.98\text{ V}$ kundrejt $5.59\text{ V}$ për të njëjtin $V_{in} = 11.28\text{ V}$).
* **Pasiguria e matjes dhe fushat e pashënuara (`N/A`):** Disa kolona (si `R2/R3` dhe `Vin (V)` tek Eksperimenti 3) nuk janë të zbatueshme ose nuk janë regjistruar (`N/A`), ndërsa vlerat e matura në Eksperimentin 3 ($0.1600\text{ V}$ deri në $3.4783\text{ V}$) mbajnë katër shifra pas presjes dhjetore sipas leximeve të multimetrit.

## 6. Diskutimi
* **Krahasimi mes kΩ dhe MΩ (Eksperimenti 1 vs. Eksperimenti 2):** Me të njëjtin tension hyrës $V_{in} = 11.28\text{ V}$, rezistorët në rangun $\text{k}\Omega$ ($2.22\text{ k}\Omega$ dhe $2.18\text{ k}\Omega$) japin $V_{out} = 5.59\text{ V}$ dhe raport $U_{out}/U_{in} = 0.4956$ (shumë afër $0.5$). Kur kalohet te rezistorët në rangun $\text{M}\Omega$ ($3.212\text{ M}\Omega$ dhe $3.250\text{ M}\Omega$), $V_{out}$ ulet në $4.98\text{ V}$ dhe raporti zbret në $0.4415$.
* **Ndikimi i ngarkesës R3 / RL (Eksperimenti 3):** Duke rritur vlerën e $R_3 / R_L$ nga $46\,\Omega$ në $100\,\Omega$, $220\,\Omega$, $470\,\Omega$ dhe $1000\,\Omega$, tensioni i daljes rritet vazhdimisht nga $0.1600\text{ V}$ në $0.3480\text{ V}$, $0.7656\text{ V}$, $1.6786\text{ V}$ dhe $3.4783\text{ V}$, duke treguar rritje të tensionit të ngarkesës me rritjen e rezistencës së saj.
